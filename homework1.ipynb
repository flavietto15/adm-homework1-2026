## alphabet-rangoli
https://www.hackerrank.com/challenges/alphabet-rangoli/problem

```python
# https://www.hackerrank.com/challenges/alphabet-rangoli/problem
from string import ascii_lowercase

def print_rangoli(size):
    rows = []
    for i in range(size):
        left = ascii_lowercase[size - 1:i:-1]
        row = '-'.join(left + ascii_lowercase[i] + left[::-1])
        rows.append(row.center(4 * size - 3, '-'))
    print(*rows[:0:-1], *rows, sep='\n')

if __name__ == '__main__':
    print_rangoli(int(input()))
```

## the-minion-game
https://www.hackerrank.com/challenges/the-minion-game/problem

```python
# https://www.hackerrank.com/challenges/the-minion-game/problem
def minion_game(string):
    kevin = sum(len(string) - i for i, c in enumerate(string) if c in 'AEIOU')
    stuart = len(string) * (len(string) + 1) // 2 - kevin
    if kevin == stuart:
        print('Draw')
    elif kevin > stuart:
        print('Kevin', kevin)
    else:
        print('Stuart', stuart)

if __name__ == '__main__':
    minion_game(input())
```

## piling-up
https://www.hackerrank.com/challenges/piling-up/problem

```python
# https://www.hackerrank.com/challenges/piling-up/problem
from collections import deque
for _ in range(int(input())):
    input()
    blocks = deque(map(int, input().split()))
    previous = float('inf')
    possible = True
    while blocks:
        current = blocks.popleft() if blocks[0] >= blocks[-1] else blocks.pop()
        if current > previous:
            possible = False
            break
        previous = current
    print('Yes' if possible else 'No')
```

## map-and-lambda-expression

```python
# https://www.hackerrank.com/challenges/map-and-lambda-expression/problem
cube = lambda x: x ** 3

def fibonacci(n):
    result = []
    a, b = 0, 1
    for _ in range(n):
        result.append(a)
        a, b = b, a + b
    return result

if __name__ == '__main__':
    print(list(map(cube, fibonacci(int(input())))))
```

```python
# https://www.hackerrank.com/challenges/html-parser-part-1/problem
from html.parser import HTMLParser

class Parser(HTMLParser):
    def handle_starttag(self, tag, attrs):
        print('Start :', tag)
        self.attributes(attrs)

    def handle_endtag(self, tag):
        print('End   :', tag)

    def handle_startendtag(self, tag, attrs):
        print('Empty :', tag)
        self.attributes(attrs)

    def attributes(self, attrs):
        for name, value in attrs:
            print('->', name, '>', value)

Parser().feed('\n'.join(input() for _ in range(int(input()))))
```

```python
# https://www.hackerrank.com/challenges/html-parser-part-2/problem
from html.parser import HTMLParser

class Parser(HTMLParser):
    def handle_comment(self, data):
        print('>>> Multi-line Comment' if '\n' in data else '>>> Single-line Comment')
        print(data)

    def handle_data(self, data):
        if data != '\n':
            print('>>> Data')
            print(data)

Parser().feed('\n'.join(input() for _ in range(int(input()))))
```

## validating-uid
https://www.hackerrank.com/challenges/validating-uid/problem

```python
# https://www.hackerrank.com/challenges/validating-uid/problem
import re
for _ in range(int(input())):
    uid = input()
    valid = (re.fullmatch(r'[A-Za-z0-9]{10}', uid) is not None
             and len(set(uid)) == 10
             and len(re.findall(r'[A-Z]', uid)) >= 2
             and len(re.findall(r'[0-9]', uid)) >= 3)
    print('Valid' if valid else 'Invalid')
```

## validating-credit-card-number
https://www.hackerrank.com/challenges/validating-credit-card-number/problem

```python
# https://www.hackerrank.com/challenges/validating-credit-card-number/problem
import re
for _ in range(int(input())):
    card = input()
    valid_format = re.fullmatch(r'[456][0-9]{15}|[456][0-9]{3}(?:-[0-9]{4}){3}', card)
    repeats = re.search(r'([0-9])\1{3}', card.replace('-', ''))
    print('Valid' if valid_format and not repeats else 'Invalid')
```

## xml-1-find-the-score

```python
# https://www.hackerrank.com/challenges/xml-1-find-the-score/problem
import xml.etree.ElementTree as ET

def get_attr_number(node):
    return sum(len(element.attrib) for element in node.iter())

if __name__ == '__main__':
    root = ET.fromstring('\n'.join(input() for _ in range(int(input()))))
    print(get_attr_number(root))
```

## xml2-find-the-maximum-depth
https://www.hackerrank.com/challenges/xml2-find-the-maximum-depth/problem

```python
# https://www.hackerrank.com/challenges/xml2-find-the-maximum-depth/problem
import xml.etree.ElementTree as ET
maxdepth = 0

def depth(elem, level):
    global maxdepth
    level += 1
    maxdepth = max(maxdepth, level)
    for child in elem:
        depth(child, level)

if __name__ == '__main__':
    root = ET.fromstring('\n'.join(input() for _ in range(int(input()))))
    depth(root, -1)
    print(maxdepth)
```

## standardize-mobile-number-using-decorators
https://www.hackerrank.com/challenges/standardize-mobile-number-using-decorators/problem

```python
# https://www.hackerrank.com/challenges/standardize-mobile-number-using-decorators/problem
def wrapper(f):
    def fun(numbers):
        normalized = ['+91 ' + number[-10:-5] + ' ' + number[-5:] for number in numbers]
        return f(normalized)
    return fun

@wrapper
def sort_phone(numbers):
    print(*sorted(numbers), sep='\n')

if __name__ == '__main__':
    sort_phone([input() for _ in range(int(input()))])
```

## decorators-2-name-directory
https://www.hackerrank.com/challenges/decorators-2-name-directory/problem

```python
# https://www.hackerrank.com/challenges/decorators-2-name-directory/problem
def person_lister(f):
    def inner(people):
        return [f(person) for person in sorted(people, key=lambda p: int(p[2]))]
    return inner

@person_lister
def name_format(person):
    return ('Mr. ' if person[3] == 'M' else 'Ms. ') + person[0] + ' ' + person[1]

if __name__ == '__main__':
    people = [input().split() for _ in range(int(input()))]
    print(*name_format(people), sep='\n')
```

## kangaroo
https://www.hackerrank.com/challenges/kangaroo/problem

```python
# https://www.hackerrank.com/challenges/kangaroo/problem
def kangaroo(x1, v1, x2, v2):
    distance, speed = x2 - x1, v1 - v2
    if speed == 0:
        return 'YES' if distance == 0 else 'NO'
    return 'YES' if distance % speed == 0 and distance // speed >= 0 else 'NO'

if __name__ == '__main__':
    print(kangaroo(*map(int, input().split())))
```

## strange-advertising
https://www.hackerrank.com/challenges/strange-advertising/problem

```python
# https://www.hackerrank.com/challenges/strange-advertising/problem
def viralAdvertising(n):
    shared, total = 5, 0
    for _ in range(n):
        liked = shared // 2
        total += liked
        shared = liked * 3
    return total

if __name__ == '__main__':
    print(viralAdvertising(int(input())))
```

## recursive-digit-sum
https://www.hackerrank.com/challenges/recursive-digit-sum/problem

```python
# https://www.hackerrank.com/challenges/recursive-digit-sum/problem
def superDigit(n, k):
    total = sum(map(int, n)) * k
    while total >= 10:
        total = sum(map(int, str(total)))
    return total

if __name__ == '__main__':
    n, k = input().split()
    print(superDigit(n, int(k)))
```

## insertionsort1
https://www.hackerrank.com/challenges/insertionsort1/problem

```python
# https://www.hackerrank.com/challenges/insertionsort1/problem
def insertionSort1(n, arr):
    value = arr[-1]
    i = n - 2
    while i >= 0 and arr[i] > value:
        arr[i + 1] = arr[i]
        print(*arr)
        i -= 1
    arr[i + 1] = value
    print(*arr)

if __name__ == '__main__':
    n = int(input())
    insertionSort1(n, list(map(int, input().split())))
```

## insertionsort2
https://www.hackerrank.com/challenges/insertionsort2/problem

```python
# https://www.hackerrank.com/challenges/insertionsort2/problem
def insertionSort2(n, arr):
    for j in range(1, n):
        value = arr[j]
        i = j - 1
        while i >= 0 and arr[i] > value:
            arr[i + 1] = arr[i]
            i -= 1
        arr[i + 1] = value
        print(*arr)

if __name__ == '__main__':
    n = int(input())
    insertionSort2(n, list(map(int, input().split())))
```
